````python
import json
from google.colab import files
import os

# ========== НАСТРОЙКИ ПУТЕЙ ДЛЯ COLAB ==========
INPUT_FILENAME = "50 диалогов.txt"
OUTPUT_FILENAME = "analyst_dataset.jsonl"
INPUT_PATH = f"/media/{INPUT_FILENAME}"
OUTPUT_PATH = f"/media/{OUTPUT_FILENAME}"

# Инструкция для аналитика (фиксированная)
INSTRUCTION = """[ROLE: analyst]

Assess if a Lua generation task is specific enough to act on.
CLEAR = names field paths (wf.vars.*), the operation, and expected result.
UNCLEAR = missing field names, data shape, or transformation logic.
If CLEAR: respond PROCEED. If UNCLEAR: ask 1-3 questions. No preamble."""

# ========== ЗАГРУЗКА ФАЙЛА (если его нет) ==========
if not os.path.exists(INPUT_PATH):
    print(f"Файл {INPUT_FILENAME} не найден. Пожалуйста, загрузите его:")
    uploaded = files.upload()
    for fname in uploaded.keys():
        os.rename(fname, INPUT_PATH)
        print(f"Файл сохранён как {INPUT_PATH}")

# ========== ЗАГРУЗКА ДИАЛОГОВ ==========
with open(INPUT_PATH, "r", encoding="utf-8") as f:
    data = json.load(f)

# Приведение к единому формату: список диалогов, каждый диалог — список сообщений {role, content}
if isinstance(data, list) and all(isinstance(d, dict) and "dialog" in d for d in data):
    dialogs = [item["dialog"] for item in data]
elif isinstance(data, list) and all(isinstance(d, list) for d in data):
    dialogs = data
else:
    raise ValueError("Неизвестный формат. Ожидается массив с ключом 'dialog' или массив списков сообщений.")

print(f"Загружено диалогов: {len(dialogs)}")

# ========== ПОСТРОЕНИЕ ПРИМЕРОВ ДЛЯ АНАЛИТИКА ==========
all_examples = []
for dialog_idx, dialog in enumerate(dialogs):
    # Находим первое сообщение пользователя (нечёткая задача)
    user_messages = [msg for msg in dialog if msg["role"] == "user"]
    assistant_messages = [msg for msg in dialog if msg["role"] == "assistant"]

    if not user_messages or not assistant_messages:
        print(f"Диалог {dialog_idx+1} пропущен: нет сообщений пользователя или ассистента")
        continue

    first_user_msg = user_messages[0]["content"]
    first_assistant_msg = assistant_messages[0]["content"]  # это и есть уточняющие вопросы

    # Формируем пример для аналитика
    example = {
        "instruction": INSTRUCTION,
        "input": first_user_msg,   # нечёткая задача
        "output": first_assistant_msg  # вопросы
    }
    all_examples.append(example)

print(f"Создано примеров для аналитика: {len(all_examples)}")

# ========== СОХРАНЕНИЕ ==========
with open(OUTPUT_PATH, "w", encoding="utf-8") as f:
    for ex in all_examples:
        f.write(json.dumps(ex, ensure_ascii=False) + "\n")

print(f"Датасет сохранён в {OUTPUT_PATH}")

# ========== СКАЧИВАНИЕ ==========
files.download(OUTPUT_PATH)
````

````python
import json
import random
from google.colab import files
import os

# ==================== НАСТРОЙКИ ====================
OUTPUT_FILENAME = "critic_data.jsonl"
OUTPUT_PATH = f"/media/{OUTPUT_FILENAME}"

# Инструкция для критика
INSTRUCTION = """[ROLE: critic]

Review Lua code for the Octapi LowCode platform.
Check: wf.vars.* access (no bare names), local declarations, _utils.array.new() for new arrays, ends with return, no forbidden calls.
If correct: LGTM. If issues: list as "Line N: problem". Be concise."""

# ==================== БАНК КОРРЕКТНЫХ КОНСТРУКЦИЙ ====================
# Правильные примеры на основе вашего файла и найденных источников

# 1. Работа с таблицами / массивами (источники: [2†L6-L7], [8†L14-L15])
TABLE_EXAMPLES = {
    "sum": """local total = 0
for _, item in ipairs(wf.vars.products) do
  total = total + (item.price or 0)
end
return total""",
    "filter": """local result = _utils.array.new()
local items = wf.vars.orders
for _, item in ipairs(items) do
  if item.region == "VLG" then
    table.insert(result, item)
  end
end
return result""",
    "collect": """local result = _utils.array.new()
for _, item in ipairs(wf.vars.events) do
  table.insert(result, item.type)
end
return result""",
    "first_element": """return wf.vars.packages[1]""",
    "last_element": """return wf.vars.items[#wf.vars.items]""",
    "length": """return #wf.vars.shipments""",
    "max": """local max = nil
for _, item in ipairs(wf.vars.records) do
  if max == nil or item.total > max then
    max = item.total
  end
end
return max""",
    "create_array": """local result = _utils.array.new()
table.insert(result, "first")
table.insert(result, "second")
return result""",
}

# 2. Условные операторы (источник: [6†L5-L9])
CONDITIONAL_EXAMPLES = {
    "if_else": """if wf.vars.score >= 90 then
  return "A"
elseif wf.vars.score >= 80 then
  return "B"
else
  return "C"
end""",
    "check_nil": """if wf.vars.status ~= nil then
  return wf.vars.status
else
  return "unknown"
end""",
}

# 3. Функции (источник: [5†L5-L6])
FUNCTION_EXAMPLES = {
    "simple": """local function add(a, b)
  return a + b
end
return add(wf.vars.x, wf.vars.y)""",
    "recursive": """local function factorial(n)
  if n == 0 then
    return 1
  else
    return n * factorial(n - 1)
  end
end
return factorial(wf.vars.num)""",
}

# 4. Работа со строками (источник: [3†L19-L21])
STRING_EXAMPLES = {
    "upper": """return string.upper(wf.vars.title)""",
    "concat": """return wf.vars.status .. "_" .. wf.vars.email""",
    "substring": """return string.sub(wf.vars.date, 1, 4)""",
}

# 5. Комбинированные операции
MIXED_EXAMPLES = {
    "filter_and_map": """local result = _utils.array.new()
for _, item in ipairs(wf.vars.users) do
  if item.active then
    table.insert(result, item.name)
  end
end
return result""",
    "transform_date": """local d = wf.vars.delivery_date
return string.sub(d,1,4) .. "-" .. string.sub(d,5,6) .. "-" .. string.sub(d,7,8)""",
    "first_non_nil": """local fields = {"title", "priority", "region"}
for _, key in ipairs(fields) do
  if wf.vars[key] ~= nil then
    return wf.vars[key]
  end
end
return nil""",
}

# Объединяем все правильные примеры в один список
ALL_CORRECT = (list(TABLE_EXAMPLES.values()) +
               list(CONDITIONAL_EXAMPLES.values()) +
               list(FUNCTION_EXAMPLES.values()) +
               list(STRING_EXAMPLES.values()) +
               list(MIXED_EXAMPLES.values()))

# ==================== ФУНКЦИИ ДЛЯ ВНЕСЕНИЯ ОШИБОК ====================

def add_missing_local(code):
    """Удаляет local у первой переменной (если есть)"""
    if "local " in code:
        return code.replace("local ", "", 1)
    return code

def add_forbidden_call(code):
    """Добавляет запрещённый вызов io.open"""
    return "io.open('file.txt', 'r')\n" + code

def remove_return(code):
    """Удаляет return в конце"""
    lines = code.split('\n')
    if lines and lines[-1].startswith('return'):
        lines.pop()
        return '\n'.join(lines)
    return code

def use_global_var(code):
    """Превращает локальную переменную в глобальную"""
    if "local result" in code:
        return code.replace("local result", "result")
    if "local total" in code:
        return code.replace("local total", "total")
    return code

def remove_wf_prefix(code):
    """Убирает wf. у wf.vars"""
    return code.replace("wf.vars.", "vars.")

def misuse_array(code):
    """Использует {} вместо _utils.array.new()"""
    return code.replace("_utils.array.new()", "{}")

def add_jsonpath(code):
    """Добавляет строку с JsonPath синтаксисом"""
    return "$.items[0]\n" + code

def remove_nil_check(code):
    """Убирает проверку на nil (влияет на логику)"""
    return code.replace("~= nil", "").replace("== nil", "")

def add_trailing_comment(code):
    """Добавляет пояснительный комментарий после return (нарушает требование)"""
    return code + "\n-- Это пояснение нарушает правило"

def wrong_array_index(code):
    """Использует индекс 0 вместо 1 в table.insert"""
    return code.replace("table.insert(", "table.insert(..., 0, ")

# Список функций-мутаторов
ERROR_FUNCTIONS = [
    add_missing_local,
    add_forbidden_call,
    remove_return,
    use_global_var,
    remove_wf_prefix,
    misuse_array,
    add_jsonpath,
    remove_nil_check,
    add_trailing_comment,
    wrong_array_index,
]

def introduce_error(code):
    """Случайным образом применяет одну из ошибок"""
    error_func = random.choice(ERROR_FUNCTIONS)
    return error_func(code)

# ==================== ГЕНЕРАЦИЯ ПРИМЕРОВ ====================
print("Генерация датасета для критика...")

all_examples = []

# 1. Генерация корректных примеров (100 штук)
for i in range(100):
    # Выбираем случайный правильный пример и немного "перемешиваем" имена переменных для разнообразия
    base_code = random.choice(ALL_CORRECT)
    # Небольшая вариативность: заменяем названия массивов и полей
    variations = [
        ("products", "items"),
        ("orders", "transactions"),
        ("price", "value"),
        ("total", "sum"),
        ("type", "category"),
    ]
    code = base_code
    for old, new in random.sample(variations, random.randint(0, 2)):
        code = code.replace(old, new)

    example = {
        "instruction": INSTRUCTION,
        "input": code,
        "output": "LGTM"
    }
    all_examples.append(example)

# 2. Генерация некорректных примеров (100 штук)
for i in range(100):
    # Берем случайный правильный пример и вносим ошибку
    base_code = random.choice(ALL_CORRECT)
    # Для большего разнообразия, сначала слегка модифицируем базовый пример
    variations = [
        ("products", "items"),
        ("orders", "transactions"),
        ("price", "value"),
        ("total", "sum"),
        ("type", "category"),
    ]
    code = base_code
    for old, new in random.sample(variations, random.randint(0, 2)):
        code = code.replace(old, new)

    corrupted_code = introduce_error(code)

    # Генерируем отчет об ошибке (имитация реального вывода критика)
    error_report = "Line 1: potential issue detected"
    if "local " not in corrupted_code and "local" in base_code:
        error_report = "Line 1: variable is missing the `local` keyword — creates a global"
    elif "io.open" in corrupted_code:
        error_report = "Line 1: forbidden call 'io.open' — not allowed in Octapi Sandbox"
    elif "return" not in corrupted_code and "return" in base_code:
        error_report = "Line X: missing `return` statement at the end"
    elif "vars." in corrupted_code and "wf.vars" not in corrupted_code:
        error_report = "Line 1: variable 'vars' is not defined — use 'wf.vars'"
    elif "{}" in corrupted_code and "_utils.array.new()" in base_code:
        error_report = "Line 1: use '_utils.array.new()' to create new arrays, not '{}'"
    elif "$." in corrupted_code:
        error_report = "Line 1: JsonPath syntax is not allowed in Lua"
    elif "-- Это пояснение" in corrupted_code:
        error_report = "Line X: output must be ONLY raw Lua code — no comments or explanations"
    else:
        error_report = "Line 1: code does not follow Octapi LowCode rules"

    example = {
        "instruction": INSTRUCTION,
        "input": corrupted_code,
        "output": error_report
    }
    all_examples.append(example)

# Перемешиваем примеры, чтобы корректные и некорректные шли в случайном порядке
random.shuffle(all_examples)

print(f"Создано примеров: {len(all_examples)} (ожидалось 200)")

# ==================== СОХРАНЕНИЕ И ВЫГРУЗКА ====================
with open(OUTPUT_PATH, "w", encoding="utf-8") as f:
    for ex in all_examples:
        f.write(json.dumps(ex, ensure_ascii=False) + "\n")

print(f"Датасет сохранён в {OUTPUT_PATH}")
files.download(OUTPUT_PATH)
````

````python
# Импорт библиотек
from unsloth import FastLanguageModel
from datasets import load_dataset
from trl import SFTTrainer
from transformers import TrainingArguments
import torch

# 1. Загрузка базовой модели с 4-битной квантизацией
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="nuprl/MultiPLCoder-1b", # Ваша базовая модель
    max_seq_length=2048, # Максимальная длина обрабатываемой последовательности
    dtype=None,
    load_in_4bit=True, # Включаем QLoRA
)

# 2. Применение LoRA-адаптера к модели
# 2. Применение LoRA-адаптера к модели
model = FastLanguageModel.get_peft_model(
    model,
    r = 16,
    target_modules = [
        "c_attn",   # Совмещённый модуль внимания
        "c_proj",   # Выходная проекция внимания
        "c_fc",     # Первый линейный слой MLP
        # "c_proj_mlp" # Опционально, иногда присутствует
    ],
    lora_alpha = 16,
    lora_dropout = 0,
    bias = "none",
    use_gradient_checkpointing = True,
)

# 3. Загрузка и подготовка датасета
# 3. Загрузка и подготовка датасета
dataset = load_dataset('json', data_files='/media/analyst_dataset.jsonl', split='train')

# Функция форматирования, возвращающая список
def formatting_func(examples):
    texts = []
    for instruction, output in zip(examples["instruction"], examples["output"]):
        text = f"### Instruction:\n{instruction}\n\n### Response:\n{output}"
        texts.append(text)
    return texts

# 4. Настройка и запуск тренера
trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset,
    formatting_func=formatting_func,
    max_seq_length=2048,
    args=TrainingArguments(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        warmup_steps=10,
        num_train_epochs=3,
        learning_rate=2e-4,
        fp16=not torch.cuda.is_bf16_supported(),
        bf16=torch.cuda.is_bf16_supported(),
        logging_steps=1,
        optim="adamw_8bit",
        output_dir="outputs",
    ),
)
````

````python
# ============================================
# ОБУЧЕНИЕ АДАПТЕРА-АНАЛИТИКА (ПРАВИЛЬНЫЙ ФОРМАТ)
# ============================================

# --- 1. Установка зависимостей ---
!pip install -q "unsloth[colab-new] @ git+https://github.com/unslothai/unsloth.git"
!pip install -q --no-deps xformers trl peft accelerate bitsandbytes
!pip install -q datasets

# --- 2. Импорты ---
from unsloth import FastLanguageModel
from datasets import load_dataset
from trl import SFTTrainer
from transformers import TrainingArguments
import torch
import os

# --- 3. Монтирование Google Drive (чтобы сохранить адаптер) ---
from google.colab import drive
drive.mount('/content/drive')

# --- 4. Пути (ИЗМЕНИТЕ ПОД СВОИ) ---
DATASET_PATH = "/media/analyst_dataset.jsonl"   # ваш файл .jsonl
SAVE_PATH    = "/content/drive/MyDrive/lora_adapters/analyst" # куда сохранить адаптер

os.makedirs(SAVE_PATH, exist_ok=True)

# --- 5. Загрузка базовой модели MultiPLCoder-1b ---
print("Загрузка базовой модели...")
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="nuprl/MultiPLCoder-1b",
    max_seq_length=2048,
    dtype=None,
    load_in_4bit=True,
)

# --- 6. Применение LoRA с ПРАВИЛЬНЫМИ target_modules ---
print("Применение LoRA...")
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=[
        "c_attn",   # для архитектуры StarCoder (MultiPLCoder-1b)
        "c_proj",
        "c_fc",
    ],
    lora_alpha=16,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing=True,
)

# --- 7. Загрузка датасета ---
print("Загрузка датасета...")
dataset = load_dataset('json', data_files=DATASET_PATH, split='train')
print(f"Загружено примеров: {len(dataset)}")

# --- 8. Функция форматирования (исправленная!) ---
# Используем поля: 'input' (запрос пользователя) и 'output' (вопросы аналитика)
# Системную инструкцию можно добавить в начало при инференсе, в обучении она не обязательна
def formatting_func(examples):
    texts = []
    for inp, out in zip(examples["input"], examples["output"]):
        # Шаблон: User: ... \n Assistant: ...
        text = f"### User:\n{inp}\n\n### Assistant:\n{out}"
        texts.append(text)
    return texts

# --- 9. Настройка и запуск тренера ---
print("Запуск обучения...")
trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset,
    formatting_func=formatting_func,
    max_seq_length=2048,
    args=TrainingArguments(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        warmup_steps=10,
        num_train_epochs=5,                     # чуть больше эпох для маленького датасета
        learning_rate=2e-4,
        fp16=not torch.cuda.is_bf16_supported(),
        bf16=torch.cuda.is_bf16_supported(),
        logging_steps=1,
        optim="adamw_8bit",
        output_dir="/content/outputs_tmp",      # временная папка
    ),
)

trainer.train()

# --- 10. Сохранение обученного адаптера ---
print("Сохранение адаптера...")
model.save_pretrained(SAVE_PATH)
tokenizer.save_pretrained(SAVE_PATH)

print(f"✅ Адаптер сохранён в {SAVE_PATH}")
````

````python
# ============================================
# ЯЧЕЙКА ИНФЕРЕНСА С ПОСТОБРАБОТКОЙ (ДО 5 ВОПРОСОВ)
# ============================================

# --- 1. Установка библиотек (если не установлены) ---
!pip install -q unsloth peft transformers accelerate bitsandbytes

# --- 2. Импорты ---
from unsloth import FastLanguageModel
from peft import PeftModel
import torch
import re
import os
from google.colab import drive

# --- 3. Монтирование Google Drive ---
drive.mount('/content/drive')

# --- 4. Путь к вашему адаптеру (измените при необходимости) ---
ADAPTER_PATH = "/content/drive/MyDrive/lora_adapters/analyst"

# --- 5. Загрузка базовой модели MultiPLCoder-1b ---
print("Загрузка базовой модели...")
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="nuprl/MultiPLCoder-1b",
    max_seq_length=2048,
    load_in_4bit=True,
)

# --- 6. Загрузка обученного адаптера ---
print("Загрузка адаптера-аналитика...")
if not os.path.exists(ADAPTER_PATH):
    raise FileNotFoundError(f"Адаптер не найден по пути: {ADAPTER_PATH}")
model = PeftModel.from_pretrained(model, ADAPTER_PATH)
FastLanguageModel.for_inference(model)
print("Модель готова.\n")

# --- 7. Тестовый запрос (можно заменить на любой) ---
user_task = "Напиши скрипт для обработки данных на Lua."

# --- 8. Формирование промпта (как при обучении) ---
prompt = f"### User:\n{user_task}\n\n### Assistant:\n"

# --- 9. Токенизация и генерация ---
inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
outputs = model.generate(
    **inputs,
    max_new_tokens=256,
    temperature=0.1,
    do_sample=True,
    repetition_penalty=1.1
)

generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print("=== ПОЛНЫЙ ОТВЕТ МОДЕЛИ ===")
print(generated_text)
print("\n")

# --- 10. Постобработка: оставляем до 5 вопросов ---
lines = generated_text.split('\n')
questions = []
for line in lines:
    stripped = line.strip()
    # Ищем строки, которые начинаются с цифры и точки (например "1. Текст вопроса")
    if stripped and re.match(r'^\d+\.', stripped):
        questions.append(stripped)
    if len(questions) >= 5:
        break

if questions:
    # Формируем чистый ответ: первая строка (обычно заголовок) + отфильтрованные вопросы
    header = lines[0].strip() if lines else ""
    filtered_output = header + '\n' + '\n'.join(questions)
    print("=== ОТФИЛЬТРОВАННЫЙ ОТВЕТ (ДО 5 ВОПРОСОВ) ===")
    print(filtered_output)
else:
    print("⚠️ Не удалось выделить нумерованные вопросы. Используйте полный ответ.")
````

````python
# ============================================
# АГЕНТ ДЛЯ ГЕНЕРАЦИИ LUA-КОДА (АНАЛИТИК + ГЕНЕРАТОР + КРИТИК + LUACHECK)
# ============================================

# --- 1. Установка зависимостей ---
!pip install -q langgraph langchain langchain-ollama unsloth peft transformers accelerate bitsandbytes
!apt-get install -y luarocks > /dev/null 2>&1
!luarocks install luacheck > /dev/null 2>&1

# --- 2. Импорты ---
import subprocess
import json
import re
import os
from typing import TypedDict, Literal
from google.colab import drive
from unsloth import FastLanguageModel
from peft import PeftModel
import torch
from langgraph.graph import StateGraph, END

# --- 3. Монтирование Google Drive ---
drive.mount('/content/drive')

# --- 4. Пути к адаптерам (измените при необходимости) ---
ANALYST_PATH = "/content/drive/MyDrive/lora_adapters/analyst"
GENERATOR_PATH = "/content/drive/MyDrive/lora_adapters/generator"
CRITIC_PATH = "/content/drive/MyDrive/lora_adapters/critic_v2"

# --- 5. Загрузка базовой модели MultiPLCoder-1b (один раз) ---
print("⏳ Загрузка базовой модели...")
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="nuprl/MultiPLCoder-1b",
    max_seq_length=2048,
    load_in_4bit=True,
)
FastLanguageModel.for_inference(model)
print("✅ Базовая модель загружена")

# --- 6. Функция для загрузки адаптера ---
def load_adapter(adapter_path):
    m = PeftModel.from_pretrained(model, adapter_path)
    FastLanguageModel.for_inference(m)
    return m

# --- 7. Загружаем все три адаптера (переключаться будем динамически) ---
print("⏳ Загрузка адаптеров...")
model_analyst = load_adapter(ANALYST_PATH)
model_generator = load_adapter(GENERATOR_PATH)
model_critic = load_adapter(CRITIC_PATH)
print("✅ Адаптеры загружены")

# --- 8. Функция вызова Luacheck ---
def check_luacheck(code: str) -> tuple[bool, str]:
    """Возвращает (успех, сообщение_об_ошибке)."""
    result = subprocess.run(
        ['luacheck', '--formatter', 'plain', '--no-color', '-'],
        input=code, capture_output=True, text=True
    )
    if result.returncode == 0:
        return True, ""
    else:
        return False, result.stderr.strip() or result.stdout.strip()

# --- 9. Определение состояния агента ---
class AgentState(TypedDict):
    user_input: str          # исходный запрос пользователя
    context: str             # накопленный контекст (вопросы и ответы)
    lua_code: str            # сгенерированный код
    error_log: str           # ошибки валидации
    iteration: int           # счётчик итераций доработки
    need_clarify: bool       # требуется ли уточнение
    next_step: Literal["analyst", "generator", "critic", "luacheck", "end", "ask_user"]

# --- 10. Узлы графа ---

def analyst_node(state: AgentState) -> AgentState:
    """Адаптер-аналитик задаёт уточняющие вопросы."""
    prompt = f"### User:\n{state['user_input']}\n\n### Assistant:\n"
    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
    outputs = model_analyst.generate(
        **inputs, max_new_tokens=256, temperature=0.2, do_sample=True, repetition_penalty=1.1
    )
    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    # Извлекаем только часть после "### Assistant:"
    if "### Assistant:" in response:
        response = response.split("### Assistant:")[-1].strip()
    state["context"] = response
    # Определяем, есть ли в ответе вопросы (начинающиеся с цифры или содержащие '?')
    questions = re.findall(r"\d+\.\s", response)
    if len(questions) >= 1:
        state["need_clarify"] = True
        state["next_step"] = "ask_user"
    else:
        state["need_clarify"] = False
        state["next_step"] = "generator"
    return state

def ask_user_node(state: AgentState) -> AgentState:
    """Выводит вопросы и ожидает ответа от пользователя (в реальном времени)."""
    print("\n🤖 Агент хочет уточнить:\n")
    print(state["context"])
    answers = input("\n✏️ Ваш ответ (можно кратко): ")
    state["user_input"] = f"{state['user_input']}\n\nУточнения: {answers}"
    state["need_clarify"] = False
    state["next_step"] = "generator"
    return state

def generator_node(state: AgentState) -> AgentState:
    """Генерирует код на основе уточнённого запроса."""
    prompt = f"### User:\n{state['user_input']}\n\n### Assistant:\n"
    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
    outputs = model_generator.generate(
        **inputs, max_new_tokens=512, temperature=0.2, do_sample=True, repetition_penalty=1.1
    )
    code = tokenizer.decode(outputs[0], skip_special_tokens=True)
    if "### Assistant:" in code:
        code = code.split("### Assistant:")[-1].strip()
    # Убираем маркеры кода ```lua ... ```
    code = re.sub(r"```lua\n?|```", "", code).strip()
    state["lua_code"] = code
    state["iteration"] = state.get("iteration", 0) + 1
    state["next_step"] = "luacheck"
    return state

def luacheck_node(state: AgentState) -> AgentState:
    """Проверяет синтаксис через Luacheck."""
    ok, errors = check_luacheck(state["lua_code"])
    if ok:
        state["next_step"] = "critic"
    else:
        state["error_log"] = f"Синтаксические ошибки:\n{errors}"
        if state["iteration"] < 3:
            state["next_step"] = "generator"  # отправить на доработку
        else:
            state["next_step"] = "end"
    return state

def critic_node(state: AgentState) -> AgentState:
    """Адаптер-критик анализирует логику кода."""
    prompt = f"### Review the logic of this Lua code:\n{state['lua_code']}\n\n### Analysis:\n"
    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
    outputs = model_critic.generate(
        **inputs, max_new_tokens=256, temperature=0.2, do_sample=True, repetition_penalty=1.1
    )
    analysis = tokenizer.decode(outputs[0], skip_special_tokens=True)
    if "### Analysis:" in analysis:
        analysis = analysis.split("### Analysis:")[-1].strip()
    # Если критик нашёл серьёзные проблемы, отправляем на доработку
    if any(word in analysis.lower() for word in ["ошибка", "проблема", "неверно", "error", "incorrect"]):
        state["error_log"] = f"Замечания по логике:\n{analysis}"
        if state["iteration"] < 3:
            state["next_step"] = "generator"
        else:
            state["next_step"] = "end"
    else:
        state["error_log"] = ""
        state["next_step"] = "end"
    return state

def should_continue(state: AgentState) -> str:
    """Определяет, куда идти дальше."""
    return state["next_step"]

# --- 11. Построение графа ---
workflow = StateGraph(AgentState)

workflow.add_node("analyst", analyst_node)
workflow.add_node("ask_user", ask_user_node)
workflow.add_node("generator", generator_node)
workflow.add_node("luacheck", luacheck_node)
workflow.add_node("critic", critic_node)

workflow.set_entry_point("analyst")
workflow.add_edge("analyst", "ask_user")
workflow.add_edge("ask_user", "generator")
workflow.add_edge("generator", "luacheck")
workflow.add_edge("luacheck", "critic")
workflow.add_conditional_edges(
    "critic",
    should_continue,
    {
        "generator": "generator",
        "end": END,
        "luacheck": "luacheck"
    }
)

app = workflow.compile()

# --- 12. Интерактивный запуск ---
print("\n" + "="*50)
print("🚀 АГЕНТ ГОТОВ К РАБОТЕ")
print("="*50)

while True:
    user_task = input("\n📝 Введите задачу (или 'exit' для выхода): ")
    if user_task.lower() == 'exit':
        break

    state = {
        "user_input": user_task,
        "context": "",
        "lua_code": "",
        "error_log": "",
        "iteration": 0,
        "need_clarify": False,
        "next_step": "analyst"
    }

    final_state = app.invoke(state)

    if final_state.get("lua_code"):
        print("\n" + "="*50)
        print("✅ ИТОГОВЫЙ КОД LUA:")
        print("="*50)
        print(final_state["lua_code"])
    else:
        print("\n❌ Не удалось сгенерировать код. Попробуйте уточнить задачу.")
````

````python
# ============================================
# ФИНАЛЬНЫЙ АГЕНТ НА QWEN2.5-CODER 1.5B (OLLAMA)
# ============================================

# --- 1. Установка LangGraph и Luacheck ---
!pip install -q langgraph langchain langchain-ollama
!apt-get install -y luarocks > /dev/null 2>&1
!luarocks install luacheck > /dev/null 2>&1

# --- 2. Импорты ---
import subprocess
import re
from typing import TypedDict, Literal
from langgraph.graph import StateGraph, END
from langchain_ollama import ChatOllama

# --- 3. Инициализация модели Qwen через Ollama ---
# Используем кэшированную модель, загруженную ранее
llm = ChatOllama(
    model="qwen2.5-coder:1.5b-instruct-q5_K_M",
    num_ctx=4096,
    num_predict=512,
    temperature=0.2,
    repeat_penalty=1.1
)

def check_luacheck(code: str) -> tuple[bool, str]:
    """Проверяет синтаксис Lua-кода через Luacheck."""
    result = subprocess.run(
        ['luacheck', '--formatter', 'plain', '--no-color', '-'],
        input=code, capture_output=True, text=True
    )
    return (True, "") if result.returncode == 0 else (False, result.stderr.strip() or result.stdout.strip())

# --- 4. Состояние агента ---
class AgentState(TypedDict):
    user_input: str
    lua_code: str
    error_log: str
    iteration: int
    need_clarify: bool
    clarify_questions: str
    next_step: Literal["generator", "critic", "luacheck", "ask_user", "end"]

# --- 5. Узлы графа ---
def generator_node(state: AgentState) -> AgentState:
    """Генерирует код или задаёт уточняющие вопросы."""
    system_prompt = """Ты — ассистент для написания кода на Lua.
Если задача пользователя неполная (не указаны детали, формат данных, источник, цель), сначала задай 2-4 конкретных уточняющих вопроса. Начинай ответ со слова "ВОПРОСЫ:", если нужны уточнения.
Если данных достаточно, сразу напиши код и начни ответ со слова "КОД:".
Не добавляй лишних комментариев вне кода или вопросов."""

    messages = [
        ("system", system_prompt),
        ("human", state['user_input'])
    ]
    response = llm.invoke(messages).content.strip()

    if response.startswith("ВОПРОСЫ:"):
        state["need_clarify"] = True
        state["clarify_questions"] = response
        state["next_step"] = "ask_user"
    else:
        state["need_clarify"] = False
        # Убираем префикс "КОД:" если есть
        code = response[4:].strip() if response.startswith("КОД:") else response
        # Удаляем маркеры Markdown ```lua ... ```
        code = re.sub(r"```lua\n?|```", "", code).strip()
        state["lua_code"] = code
        state["iteration"] = state.get("iteration", 0) + 1
        state["next_step"] = "luacheck"
    return state

def ask_user_node(state: AgentState) -> AgentState:
    """Выводит вопросы и ожидает ответа от пользователя."""
    print("\n🤖 Агент хочет уточнить:\n")
    print(state["clarify_questions"])
    answers = input("\n✏️ Ваш ответ: ")
    state["user_input"] = f"{state['user_input']}\n\nУточнения: {answers}"
    state["need_clarify"] = False
    state["next_step"] = "generator"
    return state

def luacheck_node(state: AgentState) -> AgentState:
    """Проверяет синтаксис через Luacheck."""
    ok, errors = check_luacheck(state["lua_code"])
    if ok:
        state["next_step"] = "critic"
    else:
        state["error_log"] = f"Синтаксические ошибки:\n{errors}"
        state["next_step"] = "generator" if state["iteration"] < 3 else "end"
    return state

def critic_node(state: AgentState) -> AgentState:
    """Анализирует логику кода."""
    prompt = f"### Review the logic of this Lua code:\n{state['lua_code']}\n\n### Analysis:\n"
    # Используем raw вызов, так как критик не обучен на Qwen, но Qwen справится с zero-shot
    response = llm.invoke(prompt).content.strip()
    if any(w in response.lower() for w in ["ошибка", "проблема", "неверно", "error", "incorrect"]):
        state["error_log"] = f"Замечания по логике:\n{response}"
        state["next_step"] = "generator" if state["iteration"] < 3 else "end"
    else:
        state["error_log"] = ""
        state["next_step"] = "end"
    return state

def should_continue(state: AgentState) -> str:
    return state["next_step"]

# --- 6. Построение графа ---
workflow = StateGraph(AgentState)
workflow.add_node("generator", generator_node)
workflow.add_node("ask_user", ask_user_node)
workflow.add_node("luacheck", luacheck_node)
workflow.add_node("critic", critic_node)

workflow.set_entry_point("generator")
workflow.add_conditional_edges("generator", should_continue, {"ask_user": "ask_user", "luacheck": "luacheck", "end": END})
workflow.add_edge("ask_user", "generator")
workflow.add_edge("luacheck", "critic")
workflow.add_conditional_edges("critic", should_continue, {"generator": "generator", "end": END, "luacheck": "luacheck"})

app = workflow.compile()

# --- 7. Интерактивный цикл ---
print("\n" + "="*50)
print("🚀 АГЕНТ НА QWEN2.5-CODER ГОТОВ")
print("="*50)

while True:
    user_task = input("\n📝 Введите задачу (или 'exit'): ")
    if user_task.lower() == 'exit':
        break

    state = {
        "user_input": user_task,
        "lua_code": "",
        "error_log": "",
        "iteration": 0,
        "need_clarify": False,
        "clarify_questions": "",
        "next_step": "generator"
    }

    final_state = app.invoke(state)

    if final_state.get("lua_code"):
        print("\n" + "="*50)
        print("✅ ИТОГОВЫЙ КОД LUA:")
        print("="*50)
        print(final_state["lua_code"])
    else:
        print("\n❌ Не удалось сгенерировать код.")
````
